```python
from services.genre_evolution import prepare_genre_evolution
import pandas as pd

df = pd.read_csv('data/processed/cleaned.csv')

users = df["user"].unique()
user_ranges = []

aligned_start = None
aligned_end = None

for user in users:

    user_df = df[df["user"] == user]

    if user == "Maribel":
        user_df = user_df[user_df["month"] != "2015-08"]

    start = user_df["month"].min()
    end = user_df["month"].max()

    user_ranges.append((start, end))

aligned_start = max(start for start, _ in user_ranges)
aligned_end = min(end for _, end in user_ranges)

aligned_start, aligned_end
```
