```python
import itertools
import json
import pandas as pd
from services.overlap_service import compute_overlaps

df = pd.read_csv('data/processed/cleaned.csv')
users = ["Ashanti", "Gabi", "Maribel"]

results = []

for r in [2, 3]:

    for combo in itertools.combinations(users, r):

        combo_df = df[
            df["user"].isin(combo)
        ]

        overlap = compute_overlaps(combo_df)

        results.append(overlap)

with open("data/user_overlaps.json", "w") as f:

    json.dump(results, f, indent=2)
```

```python
from services.shared_artists import compute_shared_artists

results = []

for r in [2, 3]:

    for combo in itertools.combinations(users, r):

        combo_df = df[
            df["user"].isin(combo)
        ]

        shared = compute_shared_artists(combo_df)

        results.append({

            "users": list(combo),

            "shared_artists": (
                shared
                .rename(columns={
                    "master_metadata_album_artist_name": "artist"
                })
                .to_dict(orient="records")
            )
        })

with open("data/shared_artists.json", "w") as f:

    json.dump(results, f, indent=2)
```
